```python
import pandas as pd
import json
import requests
from pathlib import Path
```

```python
def find_swan_species():
    
    # найдем род Cygnus
    taxa_url = "https://api.inaturalist.org/v1/taxa"
    params = {"q": "Cygnus", "rank": "genus"}
    
    try:
        response = requests.get(taxa_url, params=params)
        
        if response.status_code == 200:
            data = response.json()
            
            if data["results"]:
                cygnus_genus = data["results"][0]
                cygnus_id = cygnus_genus["id"]
                
                print(f"Найден род Cygnus: ID={cygnus_id}")
                print(f"Название: {cygnus_genus.get('preferred_common_name')}")
                
                # ищем виды в этом роде
                params = {
                    "taxon_id": cygnus_id,
                    "rank": "species",
                    "per_page": 50
                }
                
                response = requests.get(taxa_url, params=params)
                
                if response.status_code == 200:
                    species_data = response.json()
                    
                    print(f"\nНайдено видов в роде Cygnus: {len(species_data['results'])}")
                    
                    for taxon in species_data["results"]:
                        taxon_id = taxon["id"]
                        name = taxon.get("name", "")
                        
                        if taxon_id not in found_species:
                            found_species[taxon_id] = {
                                "name": name,
                                "common_name": taxon.get("preferred_common_name", ""),
                                "rank": taxon.get("rank", ""),
                                "matched_query": "cygnus_genus"
                            }
                            
                            print(f" {name} - ID: {taxon_id}")
        
    except Exception as e:
        print(f"Ошибка при поиске рода: {e}")
    
    return found_species

found_species = find_swan_species()

if found_species:
    for i, (taxon_id, info) in enumerate(found_species.items(), 1):
        print(f"\n{i}. {info['name']}")
        print(f"   ID: {taxon_id}")
        print(f"   Название: {info['common_name']}")
        print(f"   Ранг: {info['rank']}")
        print(f"   Найден по: '{info['matched_query']}'")
    
    # Сохраняем результаты
    Path("data").mkdir(exist_ok=True)
    
    with open('data/found_swan_species.json', 'w', encoding='utf-8') as f:
        json.dump(found_species, f, indent=2, ensure_ascii=False)
```

```python
print(f"Всего видов: {len(swan_species)}")

Path("data/raw/swans").mkdir(parents=True, exist_ok=True)

results = {}

# cкачиваем для каждого вида
for species_key, species_info in swan_species.items():
    taxon_id = species_info["id"]
    species_name_ru = species_info["common_name_ru"]
    
    print(f"\n {species_name_ru} (ID: {taxon_id})")

    species_dir = Path(f"data/raw/swans/{species_key}")
    species_dir.mkdir(exist_ok=True)
    
    downloaded = 0
    page = 1
    max_images = 50  # количество изображений на вид
    
    while downloaded < max_images:
        url = "https://api.inaturalist.org/v1/observations"
        params = {
            "taxon_id": taxon_id,
            "quality_grade": "research",  
            "photos": True,
            "per_page": 30,
            "page": page,
            "order_by": "observed_on",
            "order": "desc"  
        }
        
        response = requests.get(url, params=params)
        data = response.json()
        observations = data.get("results", [])
            
        if not observations:
            break
            
        for obs in observations:
                if downloaded >= max_images:
                    break
                
                if obs.get("photos"):
                    photo = obs["photos"][0]
                    photo_url = photo.get("url", "")
                    
                    if photo_url:

                        if "square" in photo_url:
                            photo_url = photo_url.replace("square", "medium")
                        
                        # cкачиваем фото
                        filename = f"{obs['id']}.jpg"
                        save_path = species_dir / filename
                        
                        if not save_path.exists():
                                img_response = requests.get(photo_url, timeout=10)
                                
                                # cохраняем
                                with open(save_path, 'wb') as f:
                                    f.write(img_response.content)
                                
                                downloaded += 1
                                print(f"{filename} ({downloaded}/{max_images})")
                                                              
        page += 1
```

```python
import shutil
from sklearn.model_selection import train_test_split

with open('data/swan_species.json', 'r', encoding='utf-8') as f:
    swan_species = json.load(f)

raw_dir = Path("data/raw/swans")
processed_dir = Path("data/processed/swans_dataset")

if processed_dir.exists():
    shutil.rmtree(processed_dir)

processed_dir.mkdir(parents=True)
(processed_dir / "train").mkdir()
(processed_dir / "val").mkdir()
(processed_dir / "test").mkdir()

all_images = []
species_counts = {}

for species_key in swan_species.keys():
    species_dir = raw_dir / species_key
    
    if species_dir.exists():
        image_files = list(species_dir.glob("*.jpg"))
        
        if image_files:
            species_counts[species_key] = len(image_files)
            species_name_ru = swan_species[species_key]["common_name_ru"]
            
            for img_path in image_files:
                all_images.append({
                    "path": str(img_path),
                    "species_key": species_key,
                    "species_name_ru": species_name_ru,
                    "filename": img_path.name,
                    "latin_name": swan_species[species_key]["latin_name"]
                })

df = pd.DataFrame(all_images)

train_df, temp_df = train_test_split(
    df, 
    test_size=0.3, 
    random_state=42, 
    stratify=df["species_key"]
)
val_df, test_df = train_test_split(
    temp_df, 
    test_size=0.5, 
    random_state=42, 
    stratify=temp_df["species_key"]
)


def copy_files(df, subset_name):
    count = 0
    subset_images = []

    for _, row in df.iterrows():
            src = Path(row["path"])
            dst_dir = processed_dir / subset_name / row["species_key"]
            dst_dir.mkdir(exist_ok=True)
            dst = dst_dir / src.name
            
            shutil.copy2(src, dst)
            
            subset_images.append({
                "species_key": row["species_key"],
                "species_name_ru": row["species_name_ru"],
                "latin_name": row["latin_name"],
                "filename": row["filename"],
                "subset": subset_name,
                "new_path": str(dst)
            })
            
            count += 1
    
    return count, subset_images

train_count, train_images = copy_files(train_df, "train")
val_count, val_images = copy_files(val_df, "val")
test_count, test_images = copy_files(test_df, "test")

train_df.to_csv(processed_dir / "train_metadata.csv", index=False, encoding='utf-8')
val_df.to_csv(processed_dir / "val_metadata.csv", index=False, encoding='utf-8')
test_df.to_csv(processed_dir / "test_metadata.csv", index=False, encoding='utf-8')

dataset_info = {
    "total_images": len(df),
    "species_count": len(swan_species),
    "split_counts": {
        "train": train_count,
        "val": val_count,
        "test": test_count
    },
    "split_ratio": {
        "train": 0.7,
        "val": 0.15,
        "test": 0.15
    },
    "species_info": swan_species,
    "species_counts": species_counts,
    "images": train_images + val_images + test_images
}

with open(processed_dir / "dataset_info.json", "w", encoding='utf-8') as f:
    json.dump(dataset_info, f, indent=2, ensure_ascii=False)
```
